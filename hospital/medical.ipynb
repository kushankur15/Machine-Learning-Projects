```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import numpy as np
%matplotlib inline
```

```python
filename = 'dirty_v3_path.csv'
df = pd.read_csv(filename)
df.head()
```

```python
df = df.drop(['random_notes','noise_col'],axis=1)
```

```python
df.head()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
df['Medical Condition'].value_counts()
```

```python
df = df.dropna(subset=['Medical Condition', 'Gender', 'Age'])
df.isnull().sum()
```

```python
df['Glucose'] = df['Glucose'].fillna(df['Glucose'].mean())
df['Blood Pressure'] = df['Blood Pressure'].fillna(df['Blood Pressure'].mean())
df.isnull().sum()
```

```python
# stres level and sleep hours accoridng to smoking and alcohol consumption

data_1 = df.groupby(['Smoking','Alcohol'])[['Stress Level','Sleep Hours']].mean()
data_1
```

```python
input_cols = list(df.columns)
input_cols.remove('Medical Condition')
target_col = 'Medical Condition'
input_cols
```

```python
numerical_cols = [input_cols[0]] + input_cols[2:]
cat_cols = 'Gender'
```

```python
numerical_cols
```

```python
df[numerical_cols].describe()
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler().fit(df[numerical_cols])
```

```python
from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder(sparse_output=False, handle_unknown='ignore').fit(df[[cat_cols]])

encoded_cols = list(encoder.get_feature_names_out([cat_cols]))
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(df[input_cols], df[target_col], random_state=42)
```

```python
x_train[numerical_cols] = scaler.transform(x_train[numerical_cols])
x_test[numerical_cols]  = scaler.transform(x_test[numerical_cols])

x_train.describe()
```

```python
encoded_cols
```

```python
x_train[encoded_cols] = encoder.transform(x_train[[cat_cols]])
x_test[encoded_cols] = encoder.transform(x_test[[cat_cols]])
```

```python
x_train
```

```python
X_train = x_train[numerical_cols + encoded_cols]
X_test = x_test[numerical_cols + encoded_cols]
```

```python
X_train.columns
```

```python
from sklearn.tree import DecisionTreeClassifier

model = DecisionTreeClassifier(random_state=42)

model.fit(X_train, y_train)
```

```python
model.max_features_
```

```python
model.classes_
```

```python
y_preds = model.predict(X_test)
y_pred_prob = model.predict_proba(X_test)
```

```python
print(f"First Model Accuracy: {model.score(X_test,y_test)*100:.5f} %")
```

```python
from sklearn.tree import plot_tree
plt.figure(figsize=(80,20))
plot_tree(model, feature_names=X_train.columns, max_depth=2, filled=True)
```

```python
importance_df = pd.DataFrame({
    'feature': X_train.columns,
    'importance': model.feature_importances_
}).sort_values('importance', ascending=False)
importance_df.head(10)
```

```python
plt.title('Feature Importance')
sns.barplot(data=importance_df.head(10), x='importance', y='feature')
```

```python
def max_depth_error(md):
    model = DecisionTreeClassifier(max_depth=md,random_state=42)
    model.fit(X_train,y_train)
    train_error = 1 - model.score(X_train,y_train)
    test_error = 1 - model.score(X_test,y_test)
    return {'max depth': md, 'training error': train_error, 'test error': test_error}
```

```python
errors_df = pd.DataFrame([max_depth_error(md) for md in range(1,50)])
errors_df
```

```python
fig = px.line(x=errors_df['max depth'], y=errors_df['test error'])
fig.show()
```

```python
# max_depth = 17 gives minimum error
model = DecisionTreeClassifier(max_depth=17,max_leaf_nodes=2**10,random_state=42).fit(X_train, y_train)
model.score(X_test, y_test)
```

```python
# random forest
from sklearn.ensemble import RandomForestClassifier
model = RandomForestClassifier(random_state=42, max_depth=17, max_leaf_nodes=2**10, n_jobs=-1, n_estimators=500, min_impurity_decrease=1e-7).fit(X_train, y_train)
model.score(X_test, y_test)
```
